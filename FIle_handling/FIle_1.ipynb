File handling: 
- It let's you read from and write to the files.
- Useful for saving the data permanently.

Modes:
- "r" = reads from file.
- "w" = write (overwrites existing files).
- "a" = append (adds to file).
- "x" = creating new file.

Using with open(....) automatically closes the file after use.

```python
# Writing to a file.
with open("data.txt", "w") as f:
    f.write("Hello, sir!! \n")
    f.write("How is the today's weather sir??\n")
    f.write("Let's go for a short ride.\n")
    f.write("Running today??\n")
    f.write("\t for tab/space, \n for new/next line ")

# Reading from a file.
with open("data.txt", "r") as f:
    content = f.read()
    print(content)

# Appending to a file. Adds new content at the end.
with open("data.txt", "a") as f:
    f.write("This is new data.")
```

```python
with open("data.txt", "r") as f:
    print(f.read())
```

```python
# Read all lines into a list.
with open("data.txt", "r") as f:
    lines = f.readlines()
    print(lines)

#  Write multiple lines at once.
with open("newfile.txt", "w") as f:
    f.writelines(["Line 1 \n", "Line 2 \n", "Line 3 \n", "Line 4 \n"])
```
