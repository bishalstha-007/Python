### Pandas
- It is a python library used for data analysis, data cleaning and data manipulation.

Series : One dimensional data, like a single column in a table.

```python
import pandas as pd

marks = pd.Series([80, 90, 75, 85])

print(marks)
```

DataFrame : Two dimensional data. It is like a excel spreadsheet with rows and columns.

```python
data = {
    "Name": ["Ram", "Sita", "Hari","Krishna", "Gita", "Samikshya"],
    "Age": [20, 21, 19, 22, 18, 21],
    "Marks": [85, 92, 78, 66, 75, 82], 
    "Address": ["Itahari", "Dharan", "Damak", "Biratnagar", "Inaruwa", "Tarahara"]
}

info = pd.DataFrame(data)

# Printing normal key value pairs.
print(data)

# After using dataframe.
print("\n", info)
```

Accessing columns and rows

```python
# Select one column(Name)
print(info["Name"])

# Select one row by index label.
print("\n")
print(info.loc[2])

# Selects rows by integer position.
print("\n")
print(info.iloc[0])

# Selcet a single value
print("\n", info.loc[1, "Marks"])
```

Filtering data based on conditions.

```python
# Students scoring more than 75

result = info[info["Marks"] > 75]

print(result)
```

Basics data analysis

```python
# Average mean
print(info["Marks"].mean())

# Highest 
print(info["Marks"].max())

# Lowest
print(info["Marks"].min())

# Statistical summary
print("\n")
print(info["Marks"].describe())
print("\n")
print(info.describe())
```
