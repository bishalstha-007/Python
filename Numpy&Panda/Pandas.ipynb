### Pandas
- It is a python library used for data analysis, data cleaning and data manipulation.

Series : One dimensional data, like a single column in a table.

```python
import pandas as pd

marks = pd.Series([80, 90, 75, 85])

print(marks)
```

DataFrame : Two dimensional data. It is like a excel spreadsheet with rows and columns.

```python
data = {
    "Name": ["Ram", "Sita", "Hari"],
    "Age": [20, 21, 19],
    "Marks": [85, 92, 78]
}

info = pd.DataFrame(data)
print(data)
```

Accessing columns and rows

```python
# Select one column(Name)
print(info["Name"])

# Select one row by index label.
print("\n")
print(info.loc[0])
```
