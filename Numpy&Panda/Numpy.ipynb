#### NumPy (Numerical Python)
Specially used for numeric data.  
Main features:
- Creates and manages multidimensional arrays.
- Performs mathematical and statistical operations.
- Supports matrix operations.
- Is widely used in scientific computing and machine learning.
- Faster than python lists.

```python
# Importing
import numpy as np

# Create an array, 1D
arr = np.array([10, 20, 30, 40,50, 60, 70, 80])

print(arr)

# Reshape = Changes dimension without changing the actual data.
print("\nFirst change :\n", arr.reshape(2,4))

print("\nSecond change: \n", arr.reshape(4,2))
```

```python
# Mathematical operations. Adds 5 to every elements.
add = arr + 5
print(f"value after adding 5 = {add}")

sub = arr -5
print(f"Value after subtraction = {sub}")

mul = arr * 5
print(f"Value after multiplication = {mul}")

div = arr / 5
print(f"Value after division = {div}")

sq = arr ** 2
print(f"Squared Value = {sq}")
# Mean value
mean = np.mean(arr)
print(f"Mean value = {mean}")

# Maximum value of an array.
print("Maximum value =", np.max(arr))

# Minimum value.
print("Minimum value =", np.min(arr))

# Total
print("Total =", np.sum(arr))

# Standard deviation
print("Standard deviation =", np.std(arr))
```

```python
a = np.array([[1,3,5,7,9],
              [2,4,6,8,10]])

# print(a)

for i in a:
    for j in i:
        print(j)

# Special Arrays
print(np.zeros(4))      # Arrays of zeros 
print(np.ones(4))       # Arrays of ones
print(np.arange(1,11))      # Array of number from 1 - 10
print(np.arange(0, 100 , 7))        # Array of number from zero with gap of 7(multiple of 7 less than 100)
```

Accessing array elements (indexing)
- 0 repesents the first element.
- -1 represents last element.

```python
print(a[0])
print(a[1])

print(a[0][0])
print(a[0][4])
print(a[1][4])
print(a[-1][-1])

# Number of rows and columns
print(a.shape)

# Number of dimension
print(a.ndim)

print(a[-1, -1])

# Calculate sum down each column 
print(a.sum(axis=0))

# Calculate sum across each row
print(a.sum(axis=1)) 

# Checks every element and tells either the value is True or False.
print(a > 4)

# Returns every value greater than 4.
print(a[a > 4])
```
