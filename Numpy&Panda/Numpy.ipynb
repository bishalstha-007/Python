#### NumPy (Numerical Python)
Specially used for numeric data.  
Main features:
- Creates and manages multidimensional arrays.
- Performs mathematical and statistical operations.
- Supports matrix operations.
- Is widely used in scientific computing and machine learning.
- Faster than python lists.

```python
# Importing
import numpy as np

# Create an array, 1D
arr = np.array([10, 20, 30, 40,50, 60, 70, 80])

print(arr)

# Reshape = Changes dimension without changing the actual data.
print("\nFirst change :\n", arr.reshape(2,4))

print("\nSecond change: \n", arr.reshape(4,2))
```

```python
# Mathematical operations. Adds 5 to every elements.
add = arr + 5
print(f"value after adding 5 = {add}")

sub = arr -5
print(f"Value after subtraction = {sub}")

mul = arr * 5
print(f"Value after multiplication = {mul}")

div = arr / 5
print(f"Value after division = {div}")

sq = arr ** 2
print(f"Squared Value = {sq}")
# Mean value
mean = np.mean(arr)
print(f"Mean value = {mean}")

# Maximum value of an array.
print("Maximum value =", np.max(arr))

# Minimum value.
print("Minimum value =", np.min(arr))

# Total
print("Total =", np.sum(arr))

# Standard deviation
print("Standard deviation =", np.std(arr))
```
