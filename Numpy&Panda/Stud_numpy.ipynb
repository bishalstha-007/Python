Small data analysis of students marks.

```python

import numpy as np

marks = np.array([
    [75, 80, 90],
    [60, 55, 70],
    [95, 92, 98],
    [40, 65, 50]
])

# Average mark for each student
student_avg = np.mean(marks, axis=1)

# Average mark for each subject
subject_avg = np.mean(marks, axis=0)

# Students whose average is at least 70
qualified = marks[student_avg >= 70]

# Highest mark in the dataset
highest = np.max(marks)

print("Student averages:", student_avg)
print("Subject averages:", subject_avg)
print("Qualified students:\n", qualified)
print("Highest mark:", highest)
```
