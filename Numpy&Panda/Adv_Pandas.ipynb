1. GroupBy - Grouping and aggregating data  
Aggregating: Combining separate parts into single group.

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "IT"],
    "Name": ["Ram", "Sita", "Hari", "Gita", "John"],
    "Salary": [50000, 60000, 40000, 45000, 70000]
})

# GroupBy() :- Divides data into groups and calculates statistics for each group 
result = df.groupby("Department")["Salary"]

print(result.mean())
print("\n")
print(result.describe())

# Calculating multiple statistics at once.
print("\n")
df.groupby("Department")["Salary"].agg(
    ["mean", "max", "min", "count"]
)
```

2.Filtering with multiple conditions.

```python
data = {
    "Name": ["Ram", "Sita", "Hari","Krishna", "Gita", "Samikshya"],
    "Age": [20, 21, 19, 22, 18, 21],
    "Marks": [85, 92, 78, 66, 75, 82], 
    "Address": ["Itahari", "Dharan", "Damak", "Biratnagar", "Inaruwa", "Tarahara"]
}

info = pd.DataFrame(data)
# print(info)

result = info[(info["Age"]>= 20) & (info["Marks"] >= 75)]
print(result)
```

### Merging DataFrames
merge() combines two datafrmes using a common column, similar to a SQL join.  
- inner:    Keeps only matching keys  
- left:	    Keeps every row from the left table  
- right:	Keeps every row from the right table  
- outer:	Keeps keys from both tables  

```python
students = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Name": ["Rumik", "Suman", "Sita", "Ram", "Hari"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Marks": [85, 90, 88, 75, 66]
})

result = pd.merge(students, marks, on="ID")
print(result)
```

### Pivot Tables
Summarizes data across rows and columns, mucj

```python

import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "IT", "HR"],
    "Gender": ["M", "F", "M", "F", "M", "F"],
    "Salary": [50000, 60000, 40000, 45000, 70000, 50000]
})

pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Gender",
    aggfunc="mean"
)

print(pivot)
```

### Handling missing data
So instead of removing all incomplete rows, you can fill missing values using statistics or other rules.

```python

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ram", "Sita", "Hari", "Gita"],
    "Age": [20, np.nan, 22, 21],
    "Salary": [30000, 40000, np.nan, 50000]
})

# Fill missing ages with the median age
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing salaries with the mean salary
df["Salary"] = df["Salary"].fillna(
    df["Salary"].mean()
)

age_md = df["Age"].median()
salary_md = df["Salary"].median()

print("MEDIAN AGE:", age_md)
print("MEDIAN SALARY", salary_md)
print("\n")

print(df)
```

### APPLY, MAP AND TRANSFORM
Let's you customize calculations and transformations.  
1. map() : Transform individual values  

```python
df = pd.DataFrame({
    "Name" : ["Ram", "Sita", "Hari"],
    "Marks" : [85, 45, 70]
})

df["result"] = df["Marks"].map(lambda x:"pass" if x > 50 else "Fail"
                               )

print(df)
```

### Sorting and ranking
- method= "dense" gives tied scores the same rank withou't skipping the next rank.
- Other ranking methods include average, min, max, and first.

```python

df = pd.DataFrame({
    "Name": ["Ram", "Sita", "Hari", "Gita"],
    "Marks": [85, 95, 70, 95]
})

# Sort by marks, highest first
print(df.sort_values("Marks", ascending=False))
print("\n")

# Rank marks
df["Rank"] = df["Marks"].rank(
    ascending=False,
    method="dense"
)

print(df)
```

### Working with dates and times
- pd.to_datetime() : converts values into datetime objects.
- .dt.month : extracts the month.
- .dt.year : extracts the year.
- .dt.day_name : () returns the weekday name.

```python

df = pd.DataFrame({
    "Date": ["2026-01-05", "2026-02-10", "2026-03-15"],
    "Sales": [1000, 1500, 2000]
})

df["Date"] = pd.to_datetime(df["Date"])

# Separating all the things from date.
df["Month"] = df["Date"].dt.month
df["Year"] = df["Date"].dt.year
df["DayName"] = df["Date"].dt.day_name()

print(df)
```

### Concatenating multiple DataFrames
- pd.concat(): Combines DataFrames vertically or horizontally.
- ignore_index=True creates a fresh index from 0 onward.
- By default, concatenation is vertical (axis=0).
- Use axis=1 to combine columns side by side.

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Ram", "Sita"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Hari", "Gita"],
    "Marks": [70, 85]
})

result = pd.concat([df1, df2], ignore_index=True)

print(result)
```

### Rolling calculations.
- Useful for detecting trends and smoothing fluctuations in time-series data.

```python


df = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Sales": [100, 120, 90, 150, 140]
})

df["MovingAverage"] = (
    df["Sales"].rolling(window=3).mean()
)

print(df)
```
