1. GroupBy - Grouping and aggregating data  
Aggregating: Combining separate parts into single group.

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "IT"],
    "Name": ["Ram", "Sita", "Hari", "Gita", "John"],
    "Salary": [50000, 60000, 40000, 45000, 70000]
})

# GroupBy() :- Divides data into groups and calculates statistics for each group 
result = df.groupby("Department")["Salary"]

print(result.mean())
print("\n")
print(result.describe())

# Calculating multiple statistics at once.
print("\n")
df.groupby("Department")["Salary"].agg(
    ["mean", "max", "min", "count"]
)
```

2.Filtering with multiple conditions.

```python
data = {
    "Name": ["Ram", "Sita", "Hari","Krishna", "Gita", "Samikshya"],
    "Age": [20, 21, 19, 22, 18, 21],
    "Marks": [85, 92, 78, 66, 75, 82], 
    "Address": ["Itahari", "Dharan", "Damak", "Biratnagar", "Inaruwa", "Tarahara"]
}

info = pd.DataFrame(data)
# print(info)

result = info[(info["Age"]>= 20) & (info["Marks"] >= 75)]
print(result)
```

### Merging DataFrames
merge() combines two datafrmes using a common column, similar to a SQL join.  
- inner:    Keeps only matching keys  
- left:	    Keeps every row from the left table  
- right:	Keeps every row from the right table  
- outer:	Keeps keys from both tables  

```python
students = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Name": ["Rumik", "Suman", "Sita", "Ram", "Hari"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Marks": [85, 90, 88, 75, 66]
})

result = pd.merge(students, marks, on="ID")
print(result)
```

### Pivot Tables
Summarizes data across rows and columns, mucj

```python

import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "IT", "HR"],
    "Gender": ["M", "F", "M", "F", "M", "F"],
    "Salary": [50000, 60000, 40000, 45000, 70000, 50000]
})

pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    columns="Gender",
    aggfunc="mean"
)

print(pivot)
```

### Handling missing data
So instead of removing all incomplete rows, you can fill missing values using statistics or other rules.

```python

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ram", "Sita", "Hari", "Gita"],
    "Age": [20, np.nan, 22, 21],
    "Salary": [30000, 40000, np.nan, 50000]
})

# Fill missing ages with the median age
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing salaries with the mean salary
df["Salary"] = df["Salary"].fillna(
    df["Salary"].mean()
)

age_md = df["Age"].median()
salary_md = df["Salary"].median()

print("MEDIAN AGE:", age_md)
print("MEDIAN SALARY", salary_md)
print("\n")

print(df)
```
