### Data Visualization
- Presenting data using graphs and charts.  
- Makes easier to understand trends, pattern and comparisons.

### Line Chart

```python
import matplotlib.pyplot as plt

months = [1,2,3,4]
sales = [101, 151, 121, 181]

# Plots months and sales 
plt.plot(months, sales)

# Title of the chart
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

### Bar Chart

```python
import matplotlib.pyplot as plt

students = ["Ram", "Shyam", "Hari"]
marks = [80, 70, 90]

plt.bar(students, marks)
plt.title("Studen Marks")
plt.show()
```

### Pie Chart
Here we need to give label to the chart

```python
import matplotlib.pyplot as plt

sizes = [40, 35, 25]
labels = ["A", "B", "C"]

plt.pie(sizes, labels= labels)
plt.show()
```
