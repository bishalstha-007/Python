### Seaborn
- Built on top of matplotlib.
- Easier, attractive and better for statistics.

### Scatter Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [30, 40, 50, 60, 75, 90]

sns.scatterplot(x= study_hours, y= marks)

plt.title("Study Hours and Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

### Histogram

```python
marks = [20, 25, 30, 35, 40, 40,
         45, 50, 50, 60, 70, 80]
plt.hist(marks, bins=6)
# bins=5 : Divides the data into 5 intervals.
plt.title("Student's marks.")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```
