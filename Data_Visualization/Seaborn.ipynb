### Seaborn
- Built on top of matplotlib.
- Easier, attractive and better for statistics.

### Scatter Plot
Shows relationship between two variables.

```python
import seaborn as sns
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [30, 40, 50, 60, 75, 90]

sns.scatterplot(x= study_hours, y= marks)

plt.title("Study Hours and Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
tips = sns.load_dataset("tips")
print(tips)
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.title("Tips per bill")
plt.show()
```

### Count Plot
Counts categories.

```python
# same tips data.
# Most tips
sns.countplot(x="day", data= tips)
plt.show()
```

```python
# Who often give more tips.
sns.countplot(x="sex", data=tips)
plt.show()
```

```python
sns.countplot(x="smoker", data=tips)
plt.show()
```

### Histogram

```python
marks = [20, 25, 30, 35, 40, 40,
         45, 50, 50, 60, 70, 80]
plt.hist(marks, bins=6)
# bins=5 : Divides the data into 5 intervals.
plt.title("Student's marks.")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

### Line plot
Used for showing trends over time

```python

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

sns.lineplot(x=x, y=y)
plt.show()
```

### Bar Plot

```python
import pandas as pd
data = {
    "Department":["IT","HR","Sales"],
    "Salary":[50000,40000,45000]
}
df = pd.DataFrame(data)

sns.barplot(x="Department", y="Salary", data=df)

plt.show()
```

### Box Plot

```python
sns.boxplot(y=tips["total_bill"])
plt.show()
```

### Violin Plot

```python
sns.violinplot(y=tips["total_bill"])
plt.show()
```

### Pair Plot

```python
iris = sns.load_dataset("iris")
sns.pairplot(iris)
plt.show()
print(iris)
```

### Heatmap

```python

Corr = iris.corr(numeric_only = True)


# annot means annotation.
# When annot= True, Seaborn prints the actual values inside each box of the heatmap

sns.heatmap(Corr, annot= True)
plt.show()
```

### Correlation Heatmap

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")

# cmap = Color map.
# It controls which colors are used in the heatmap.
```
