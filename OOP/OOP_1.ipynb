#### Object Oriented Programming:  
It is a way of structuring code around objects and entities.  
Makes code modular, reusable and easier to maintain.

1. Class and objects
- Class = Blueprint(like a design of a car)
- Object = Actual instance created from class(like Toyota)

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        print(f"Color of {self.brand} is {self.color} !!")

# Object creation
a = Car("Mercedes", "silver")
b = Car("BMW", "Navy Blue")

a.drive()
b.drive()
```

```python
class Bike:
    def __init__(self, company, lotno, color, manuf, cc, type ):
        self.company = company
        self.lotno = lotno
        self.color = color
        self.manuf = manuf
        self.cc = cc
        self.type = type

    def __init__(self, serviceint, mobil):
        self.serviceint = serviceint
        self.mobil = mobil

    def own(self):
        print(f"I own {self.company} bike, which is {self.cc} cc and color is {self.color} ")

    def details(self):
        print(f" Company = {self.company},\n Lot-no = {self.lotno} ,\n Color = {self.color}, \n Manufactured date = {self.manuf}, \n MotarBike cc = {self.cc}, \n MotarBike type = {self.type}")

    def service(self):
        print(f" Service your bike in interval of {self.serviceint}, \n Consumes the mobil of {self.mobil} l")

    Bike1 = Bike("Yamaha", 51, "Navy blue", "2024", 149, "sport")

    Bike1.details()
```
