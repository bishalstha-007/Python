#### Polymorphishm
- Means "many forms"
- Allows same method name to behave differently depending on the object that calls it.
- Same function name, Different result.

```python
class Dog:
    def sound(self):
        print("Woof")

class Bird:
    def sound(self):
        print("Chirp")

class Cat:
    def sound(self):
        print("Meow")

# Polymorphism in action.
for animal in [Bird(), Dog(), Cat()]:
    animal.sound()
```

```python
class Bike:
    def drive(self):
        print("It moves in two wheels.")

class Car:
    def drive(self):
        print("It moves in 4 wheels.")

class Boat:
    def drive(self):
        print("It moves in water.")

for option in [Bike(), Car(), Boat()]:
    option.drive()
```

#### Example of payment system.

```python
class payment:
    def pay(self, amount):
        raise NotImplementedError("Subclass must implement this method.")

class CashPayment(payment):
    def pay(self,amount):
        print(f"Paid {amount} in cash.")

class CardPayment(payment):
    def pay(self, amount):
        print(f"Paid {amount} using credit card.")

class MobileWaletPayment(payment):
    def pay(self, amount):
        print(f"Paid {amount} via mobile wallet.")

payments = [CashPayment(), CardPayment(), MobileWaletPayment()]

for method in payments:
    method.pay(5000)
```

🌍 Real-Life Analogy
Think of the action “pay”:

If you’re at a shop, you can pay by cash.

If you’re online, you might pay by card.

If you’re in Nepal, you might use eSewa or Khalti (mobile wallet).

Same action (pay), but the way it’s done changes depending on the method.
