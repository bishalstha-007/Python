##### Inheritance:  
It allows a class(child class) to reuse and extend the properties and methods of another class(parent class).  
It helps to avoid code duplication and make programs more organized.  

```python
# Parent class
class Animal:
    def speak(self):
        print("Animals can speak!!")
    def Eat(self):
        print("Animal eat food!!")

# Child class inherits the properties of parent class.
class Dog(Animal):
    def speak(self):
        print("Woof!!")
    def Eat(self):
        print("Meat!")

class Cat(Animal):
    def speak(self):
        print("Meow!!")
    def Eat(self):
        print("Milk")

class Horse(Animal):
    def speak(self):
        print("Whinny!!")
    def Eat(self):
        print("Grass!")

# Objects
dog = Dog() 
cat = Cat()
horse = Horse()

dog.speak()
dog.Eat()
cat.speak()
cat.Eat()
horse.speak()
horse.Eat()
```

### Real life example of school system.

```python
# Parent class
class Person():
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def introduce(self):
        print(f"Hello, It's me {self.name} and i'm {self.age} years old.")

#  Child class

class Student(Person):
    def __init__(self, name, age, grade):
        super().__init__(name, age)      # super() = calls method from parent class.
        self.grade = grade
    def study(self):
        print(f"{self.name} studies in grade {self.grade}.")

class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject
    def teach(self):
        print(f"{self.name} teaches {self.subject} subject.")

class Employee(Person):
    def __init__(self, name, age, company, salary):
        super().__init__(name, age)
        self.company = company
        self.salary = salary
    def work(self):
        print(f"{self.name} is a employee of {self.company} and earns ${self.salary}k.")

# Object creation:

s1 = Student("khemey",19, 9)
t1 = Teacher("reshma", 41, "English")
e1 = Employee("Reshu", 35, "abc", 65)

s1.introduce()
s1.study()
t1.introduce()
t1.teach()
e1.introduce()
e1.work()
```
