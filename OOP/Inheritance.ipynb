##### Inheritance:  
It allows a class(child class) to reuse and extend the properties and methods of another class(parent class).  
It helps to avoid code duplication and make programs more organized.  

```python
# Parent class
class Animal:
    def speak(self):
        print("Animals can speak!!")
    def Eat(self):
        print("Animal eat food!!")

# Child class inherits the properties of parent class.
class Dog(Animal):
    def speak(self):
        print("Woof!!")
    def Eat(self):
        print("Meat!")

class Cat(Animal):
    def speak(self):
        print("Meow!!")
    def Eat(self):
        print("Milk")

class Horse(Animal):
    def speak(self):
        print("Whinny!!")
    def Eat(self):
        print("Grass!")

# Objects
dog = Dog() 
cat = Cat()
horse = Horse()

dog.speak()
cat.speak()
horse.Eat()
```
