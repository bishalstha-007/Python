#### Encapsulation  
Wrapping data and methods into  single unit.  
Restricts direct access to some attributes using private variables (__variable).  
Provides controlled access via getter/setter method.

```python
class BankAmount:
    def __init__(self, balance):
        self.__balance= balance      # Private variable

    def deposite(self, amount):
        self.__balance += amount

    def GetBalance(self):
        return self.__balance

acc = BankAmount(30000)
acc.deposite(500)
print(acc.GetBalance())
```

#### Simple student example.  
The marks are kept inside the calss and can only be changed through a method that checks the value.  

```python
class Student:
    def __init__(self,name, marks):
        self.name = name
        self.__marks = marks

    def set_marks(self, marks):
        if 0 <= marks <= 100 :
            self.__marks = marks
        else :
            print("Marks must be between 0 and 100.")

    def show_marks(self):
        print(f"{self.name}'s marks : {self.__marks}")

student = Student("Ram", 98)

student.show_marks()
student.set_marks(96)
student.show_marks()
student.set_marks(1000)         # Rejects after checking.

```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
        # Private variable (cannot be accessed directly outside)

    # deposite method contolled access.
    def deposite(self,amount) :
        if amount > 0 :
            self.__balance += amount
            print(f"Deposited money = {amount}. New_balance = {self.__balance}")
        else:
            print("Deposited amount must be positive.")

    # Withdraw method.
    def Withdraw(self, amount):
        if amount <= self.__balance :
            self.__balance -= amount
            print(F"Withdrwa_amount = {amount}. Remaining amount = {self.__balance}")
        else:
            print("Insufficient Funds.")

    # getter method(safe method to check balance.)
    def get_balance(self):
        return self.__balance

# Using class.
account = BankAccount(5000)  # Initial balance is 5000.as_integer_ratio

account.deposite(2000)
account.Withdraw(500)
print(account.get_balance())

# Direct access is restricted.
# print(account.__balance)   THIS WILL GIVE ATTRIBUTE ERROR.
```
