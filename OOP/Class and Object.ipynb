#### Object Oriented Programming:  
It is a way of structuring code around objects and entities.  
Makes code modular, reusable and easier to maintain.

1. Class and objects
- Class = Blueprint(like a design of a car)
- Object = Actual instance created from class(like Toyota)

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        print(f"Color of {self.brand} is {self.color} !!")

# Object creation
a = Car("Mercedes", "silver")
b = Car("BMW", "Navy Blue")

a.drive()
b.drive()
```

```python
class Bike:
    def __init__(self, company, lotno, color, manuf, cc, type, serviceint, mobil):
        self.company = company
        self.lotno = lotno
        self.color = color
        self.manuf = manuf
        self.cc = cc
        self.type = type
        self.serviceint = serviceint
        self.mobil = mobil

    def details(self):
        print(f"Company = {self.company}, Lot-no = {self.lotno}, Color = {self.color}, Manufactured = {self.manuf}, CC = {self.cc}, Type = {self.type}")

    def service(self):
        print(f"Service interval = {self.serviceint}, Mobil consumption = {self.mobil} L")

Bike1 = Bike("Yamaha", 51, "Navy blue", "2024", 149, "Sport", "6 months", 1.5)
Bike1.details()
Bike1.service()
```
