#### Abstraction  
Hiding implementation details and showing only the essential features.  
Abstraction is often done using abstract classes (ABC module).  
It forces child classes to implement certain methods.  

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):     # abstract class
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car starts with a key.")
    
class Bike(Vehicle):
    def start(self):
        print("Bike starts with button or kick.")

# Creating objects

V1 = Car()
V2 = Bike()

V1.start()
V2.start()
```

```python
from abc import ABC, abstractmethod

class Person(ABC):
    @abstractmethod
    def age(self):
        pass            # Do nothing here for now, but don't give an error.

class student(Person):
    def age(self):
        print("Student is 10 years old.")

class employee(Person):
    def age(self):
        print("Employee is 25 years old.")
class Enterprenuer(Person):
    def age(self):
        print("Enterpreneur is 43 years old.")

a = student()
b = employee()
c = Enterprenuer()

a.age()
b.age()
c.age()
```
