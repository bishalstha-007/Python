## Error handling.
When error occurs, it prevents the system/program from crashing.  
### Structure of Error handling.

```python
try:
    # Risky code
except:
    # Runs if error occurs
else:
    # Runs if no error occurs
finally:
    # Always runs
```

Program with error handling, provides error when the input number is 0.

```python
number = int(input("Enter the number:"))
a = 10/number
print(a)
```

Improved code using error handling.

```python
try:
    number = int(input("Enter the number:"))
    answer = 10 / number
except ValueError:
    print("Please enter a whole number.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print(answer)
```
